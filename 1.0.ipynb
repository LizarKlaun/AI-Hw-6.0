```python
import numpy as np
from sklearn.linear_model import LinearRegression

X = np.array([[2], [4], [6], [8], [10], [12]])
Y = np.array([3.0, 5.5, 7.8, 10.2, 12.7, 15.0])

model = LinearRegression()
model.fit(X, Y)

new_screens = np.array([[5], [9]])
predicted_costs = model.predict(new_screens)

print(f"Формула: Y = {model.coef_[0]:.2f} * X + {model.intercept_:.2f}")
for screens, cost in zip(new_screens.flatten(), predicted_costs):
    print(f"Екранів: {screens} -> Оцінка: {cost:.2f} тис. $")
```
